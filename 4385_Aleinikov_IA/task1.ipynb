## Основные теоретические положения

**Актор** - базовый элемент представляющий отдельную единицу, участвующую во взаимодействии.

**Реляционная связь** - отношение взаимодействия или ассоциации между двумя акторами в сети

**Диада** -- базовая структура социальной сети. Два актора и связь между ними.

**Триаада** -- подструктура сети, состоит из трех акторов и всех потенциальных сетей между ними.

**Группа** -- подмножество акторов в сети, объединенное каким либо общим признаком, отношением, взаимодействием.

**Подгруппа** -- это подмножество акторов внутри группы или сети, обладающее определенными уникальными свойствами взаимодейтсвия

**Социальная сеть** - объект, представляющий собой совокупность конечного набора акторов и отношений определенных на них.

**Модальность** - характеристика, определяющяя количество типов акторов в сети. Бываю одномодальные сети -- все узля принадлежат одному и тому же типу. итд

**Пятичисловая сводка для анализа сетей**

1. Размер сети -- количество акторов в сети.
2. Плотнось сети -- доля имеющихся связей по отношению к максимально возможному количеству связей.

$$ den(G) = \frac{\sum_{i = 1}^g \sum_{j = 1}^g x_{ij}}{g(g-1)} $$

3. Компонента связности.
4. Диаметр сети.
5. Глобальный коэфициент кластеризаци -- мера, характеризующяя степень кластеризации(треугольности) всей сети в целом

$$d_t(G) = \frac{3\tau_{\triangle}(u)}{\tau_3(G)}$$

где $\tau_{△} $ - количество треугольников в котором в триаде наблюдаются ровно три связи.

где $\tau_3$ - количетсво треугольников в которых наблюдается 2 или 3 связи


**Локальный коэфициент кластеризации** -- мера, которая характеризует степень кластиеризации треугольника для конкретного узла в сети.

$$
cl(n_i) =
\begin{cases}
\frac{2N_i}{k_i(k_i - 1)}, k \leq 2 \\
0
\end{cases}
$$

$k_i$ - число смежных с i узлов

$N_i$ - количество ребер связаных смежных узлов между собой



**Центральность** -- количественная мера $C_A$ определяет относительную важность актора или его влияющие в сети на основе его позиции и связей с другими акторами

**Центральность по степени**  -- мера важности узла в сети, основаная на количестве его непосредственных связей с другими узлами.

$$ C_D(n_i) = deg(n_i) $$


**Нормированная центральность** по степени -- это доля от максимума возможного числа связей

$$ C_D(n_i)` = \frac{deg(n_i)}{g - 1} = \frac{C_D(n_i)}{g - 1} $$


**Центральность по близости** -- мера, определяющяя насколько узел близок ко всем остальным узлам в сети, на основе кратчайших путей.

$$C_C(n_i) = (\sum_{j = 1}^g d(n_i, n_j))^{-1}$$

**Нормированная центральность по близости**
$$C_C(n_i)` = (g - 1) C_C(n_i)$$

**Центральность по посредничеству** мера, определяющяя насколько часто узел выступает мостом на кратчайшем путях между узлами в сети.

$\sigma_{ij}$ - количество кротчайших путей от $n_i$ до $n_j$

$\sigma_{ij}(n_k)$ - количество кратчайших путей  соеди $n_i, n_j$ проходящих через $n_k$

$$ C_B(n_i) = \sum_{i \subset K} \frac{\sigma_{ik(n_i)}} {\sigma_{ik}}$$

**Нормировання центральность по посредничеству**

$$ C_b`(n_i) = \frac{2C_b(n_i)}{(g - 1)(g - 2)}$$

**Центральность по информации** -- мера важности узла в сети, основанная на его способности передавать информацию по всем возможным путям между другими узлами

$$ C_I(n_i) = \frac{1}{c_{ii} + (T - 2R)/g}$$

**Нормированная центральность по информации**

$$C_I`(n_i) = \frac{C_I(n_i)}{\sum_{i = 1}^{g} C_I(n_j)}$$


**Центральность для направленных отношения**

**Центральность по степени** для оргграфа учитывает его связи

$$C_D(n_i) = deg_0(n_i)$$

нормированность определяется аналогично.

 Центральность по близости формируется аналогично как и для не оргграфов с учетом направления. Цент не определен, только если орграф не является сильно связанным. Для решенения фек использовать только достижимых графов.

 $$C_C^* = \frac{|J_i|/(g - 1)}{\sum_{n_j \in J_i} d(n_i, n_j) / |J_i|}$$


 **Престиж** -- количестваенная мера $P_A$ влиятельности актора основанных на входных связях.

 **Престиж по степени** Мера статуса или влиятельности узла  в оргграфе оснащенной на количестве входный связей.

 $$P_D(n_i) = deg_I(n_i)$$

 **Нормированный престиж**

 $$ P`_D(n_i) = \frac{P_D(n_i)}{g - 1} $$

 **Престиж по близости** -- мера статуса узла в орграфе, учитывая долю узлов, достажимости для актора и среды $\rho$

 $$ P_P(n_i) = \frac{|I_i| / (g - 1)}{\sum_{n_j \in I_i} d(n_j, n_i) / |I_i|}$$

 $I_i$ - множество актороов достижимых для $n_i$

 **Престиж по рангу** -- мера влиятельности узла в направленной сети, учитывает не только количество входных связей, но и "авторитетность" тех,  кто эти связи проставляет.

  - Ранг актора $n_i$ является функцие ранга авторов, который бырал $n_i$ $P_R(n_i) = x_1  P_R(n_1) + ... + x_{g_i}P_R(n_g)$
  - Имеет g линейных уравнений с g неизвестными  $p = (P_R(n_1), ... , P_R(n_g))^T \implies$ существует решение

  $$p = X^Tp$$

  $$(E - X^T)p = 0$$

  Вводим параметр $a > 0$ -- затухание.
  - x - вектор столбец $\sum$ столбов социоматрицы Х.
  - Вектор рангов p находится из решения системы уравнений
  $$(\frac{1}{a}E - X^T)p = x$$

  - параметр а выбирается самостоятельнт исходя из условий

$$ \lambda_1 < \frac{1}{a} < 2 \lambda_1$$,

где $\lambda_1$ - наибольшое веществуенное собственное число социоматрицы X

**Нормированный престиж по рангу**
$$ P_R`(n_i) = \frac{P_R(n_i)}{m}$$

$$m = \sum_{k = 1}^{g - 1} a^k A_{g - 1}^k$$

Первой задачей является отделить неизолированые вершины. Для этого создана функция keep которая проверят, что сумма значений по столбцу и строке больше нуля. Это эквивалентно тому, что имеется хотя бы одна связь. Затем исходная матрица преобразуется так, что в ней остаются только те элементы, что имеют в себе хотя бы одну связь.

После этого исходная сеть преобразуется в граф. Это сделано сугубо для того чтобы в последствии использовать уже готовые функции из библиотеки. `mode=all` гласит о том, что мы учитываем все связи, и all и in. `diag = false` говорит о том, что мы игнорируем связи акторов самих в себя.

```r
keep <- rowSums(network_matrix) + colSums(network_matrix) > 0
network_matrix_unisolated <- network_matrix[keep, keep]

network_as_unisolated_graph <- graph_from_adjacency_matrix(
  network_matrix_unisolated,
  mode = "undirected",
  weighted = NULL,
  diag = FALSE
)


n <- vcount(network_as_unisolated_graph)

C_D <- degree(network_as_unisolated_graph, mode = "all") / (n - 1)
C_C <- closeness(network_as_unisolated_graph, mode = "all", normalized = TRUE)
C_B <- betweenness(network_as_unisolated_graph, directed = FALSE, normalized = TRUE)
C_I <- infocent(network_matrix_unisolated, gmode = "graph", diag = FALSE, rescale = TRUE)

res <- data.frame(
  n_i = seq_len(n),
  C_D = C_D,
  C_C = C_C,
  C_B = C_B,
  C_I = C_I
)

print(res)
```

```r
cat("Задание номер 1. Построение социоматрицы\n")
network_matrix_2 <- as.matrix(
                    read.csv("krackhardt_advice.csv",
                              comment.char = "#",
                              header       = FALSE))

cat("\nЗадание номер 2. Представить социоматрицу как список рёбер\n")
idx_2 <- which(network_matrix_2 != 0, arr.ind = TRUE)

network_edgelist_2 <- data.frame(
  from = idx_2[, 1],
  to   = idx_2[, 2]
)
print(network_edgelist_2[1:5, ], row.names = FALSE)

cat("\nЗадание номер 3. Присвоить имена столбцам и строкам социоматрицы\n")
rownames(network_matrix_2) <- colnames(network_matrix_2)

cat("\nЗадание номер 4. Построить сеть\n")
net_2 <- network(network_matrix_2,
                 directed    = TRUE,
                 matrix.type = "adjacency",
                 ignore.eval = TRUE)
print(net_2)

cat("\nЗадание номер 5. Отображение сети с помощью алгоритмов укладки Фрюхтермана—Рейнгольда и Камада—Каваи\n")
gplot(net_2, gmode = "digraph", mode = "fruchtermanreingold",
      displaylabels = TRUE, main = "Фрюхтерман Рейнгольд")
gplot(net_2, gmode = "digraph", mode = "kamadakawai",
      displaylabels = TRUE, main = "Камада Каваи")

cat("\nЗадание номер 6. Описательная сводка сети\n")
summary(net_2)

cat("\nЗадание номер 7. Пятичисловая сводка\n")
g <- network.size(net_2)

den <- sum(network_matrix_2 != 0) / (g * (g - 1))
den_prog <- network.density(net_2)

ks_prog <- sna::components(net_2, connected = "weak")
ks_prog_strong <- sna::components(net_2, connected = "strong")

diam <- igraph::diameter(
  igraph::graph_from_data_frame(network_edgelist_2, directed = TRUE),
  directed = TRUE
)

global_clusterization_coef <- sna::gtrans(net_2, mode = "digraph")

cat("Размер сети: ", g, "\n")
cat("Плотность: ", den, "\n")
cat("Плотность: ", den_prog, "\n")
cat("Слабые компоненты связности: ", ks_prog, "\n")
cat("Сильные компоненты связности: ", ks_prog_strong, "\n")
cat("Диаметр: ", diam, "\n")
cat("Глобальный коэфициент кластеризации: ", global_clusterization_coef, "\n")
```

*Выводы*

- *(b)* - Плотность, вычисленная вручную и программно, совпала и равна 0.452.
- *(c)* - Слабая и сильная компоненты связности по одной. Это значит, что из любого актора можно по направленным связям добраться до любого другого, то есть сеть сильно связна.
- *(d)* - Диаметр равен 3. Совет доходит от любого актора до любого другого не более чем за три шага, сеть довольно сильно связаная в плане кратчайших расстояний.
- *(e)* - Глобальный коэффициент кластеризации 0.664. Что говорит о том, что в сети много замкнутых триад.

Сеть советов плотная, сильно связная и сильно кластеризованная.
