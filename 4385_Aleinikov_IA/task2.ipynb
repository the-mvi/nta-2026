## Основные теоретические положения

Дан граф $G = G(N, L)$, ijfa $G_s = G_s(N_s, L_s)$ называется подграфом если $N_s \subset N$, $L_s \subset L$

Узел $n_i$ графа наезывается точкой сочленнения при удалении которого и всех смежных ему ребер увеличивается число компонент связности.

$k$-вершинный разрез -- набор узлов $\{n_{i1}, \ldots, n_{ik}\}$ необходимый для поддержания связности графа.

$k$-рёберный разрез -- наборе ребер, необходимый для поддержания связности графа.

Вершинная связности $k(G)$ - минимальное число $k$ для которого граф имеет $k$-вершинный разрез.

Рёберная связности $\lambda(G)$ - минимальное число $\lambda$ для которого граф имеет $\lambda$ реберный разрез

Клика -- максимальный полный подграф состоящий из трех и более узлов.

$k$ клика -- максимальный подграф в котором расстояние между двумя любыми вершинами не превышае $k$.

$$d(n_i, n_j) \leq k \quad \forall n_i, n_j \in N_s$$

$K$ клуб -- максимальный подграф диаметр которого не превышает $K$

$$\operatorname{diam}(G_s) \leq k$$

$k$ клан -- это к клика диаметр которой не превышает $k$.

$k$-плекс -- максимальный подграф, содержащий $g_s$ узлов, в котором каждый узел смежен не менее чем с $g_s - k$ узлами

$$\operatorname{deg}_s(n_i) \geq g_s - k \quad \forall n_i \in N_s$$

Пусть $G_s$ является к плексом с $g_s$ узлами и $k < (g_s + 2) / 2$ тогда $\operatorname{diam}(G_s) \leq 2$

$k$ ядро -- максимальный подграф в котором каждый узел смежен не менее с чем $k$ узлов в подграфе

$$\operatorname{deg}_s(n_i) \geq k \quad \forall n_i \in N_s$$

Клики в направленных связях могут быть обнаружены если рассматривать взаимные связи. При анализе направленных сетей это эквивалентно симметризации социоматрицы путём взятия минимуму элементов.

Обозначим эту матрицу $\mathbf{X} = (x_{ij})_{1 \leq i, j \leq g}$

где $x_{ij} = x_{ji} = 1$, $x_{ij} = x_{ji} = 0$

Полуцепть -- последовательно различных узлов, где все последователные пары узлов соединены дугой от первого ко второму или дугой от второго к первому. Направление не имеет значения.

Слабо $k$ связанные -- если для двух вершин существует полуцепь длины не более $k$.

Односторонние $k$ связанные если для вершин $n_i, n_j$ существует цепль длины не более к из $n_i$ в $n_j$ или цепь длины не более $k$ из $n_j$ в $n_i$

Сильно к связанные -- сли для вершин $n_i, n_j$ существует цепль длины не более к из $n_i$ в $n_j$ и цепь длины не более $k$ из $n_j$ в $n_i$

Рекурсивно к связанные если вершины $n_i, n_j$ сильно $k$ связаны и цепь от $n_i, n_j$ использует те же узлы и дуги, что и цепь от $n_j$ к $n_i$ в обратном порядке.

Слабо связанная $k$ клика это подграв в котором все узлы слабо к связаны и нет дополнительных узлов, также слабо к связаные со всему узлами подграфа.

Односторонние связанная $k$ клика это подграф в котором все узля односторонне к связаны и нет дополнительных узлов также односторонне к связаные со всеми узлами подграфа.

Сильно связная $k$ клика это подграф в котором все узлы сильно к связны и нет дополнитлеьных узлов. также сильно связной со всеми узлами подграфа.

Рекурсивно связная $k$ клика -- это подграф, в котормо все узлы рекурсивно к связны и нет дополнительных узлов, также рекурсивно к связные со всеми узлами подграфа.

Централизация группы -количественная мера $C_A$ определеяющяя относительную изменчивость или неоднородность центральности акторов в группе.

$$C_A = \frac{\sum_{i = 1}^{g} \left(C_A(n^*) - C_A(n_i)\right)}{\max\left(\sum_{i = 1}^{g}\left(C_A(n^*) - C_A(n_i)\right)\right)}$$

где $C_A(n^*) = \max_i C_A(n_i)$

Централизация можно рассматривать как дисперсию

$$S_A^2 = \frac{1}{g} \sum_{i = 1}^{g}(C_A(n_i) - \overline{C_A})^2$$

Централизация по степени

$$C_D = \frac{\sum_{i = 1}^{g} (C_D(n^*) - C_D(n_i))}{(g - 1)(g - 2)}$$

$$S_D^2 = \frac{1}{g} \sum_{i = 1}^{g}(C_D(n_i) - \overline{C_D})^2$$

Централизация по близости

$$C_C = \frac{\sum_{i = 1}^{g} (C_C'(n^*) - C_C'(n_i))}{(g - 1)(g - 2) / (2 g - 3)}$$

$$S_C^2 = \frac{1}{g} \sum_{i = 1}^{g}(C_C'(n_i) - \overline{C_C'})^2$$

Централизация по посредничеству.

$$C_B = \frac{2 \sum_{i = 1}^{g} \left(C_B(n^*) - C_B(n_i)\right)}{(g - 1)^2(g - 2)} = \frac{1}{g - 1} \sum_{i = 1}^g \left(C_B'(n^*) - C_B'(n_i)\right)$$

$$S_B^2 = \frac{1}{g} \sum_{i = 1}^{g}(C_B(n_i) - \overline{C_B})^2$$

Сети аффилированности

Сеть аффилированности -- сеть в которой участники афилированы друг с другом, на основе сотрудничества в роли в какой то группе или совместного в участия в определенном событии.

Свойства
1. является бимодальными сетями.
2. Связи между акторами основываются на связи участника в группе или событии
3. Позволяют изучать взаимодействие акторов в группе или событиях или событий для акторов.

Сети аффилированности состоят из двух наборов.

$N = \{n_1, \ldots, n_g\}$ набор акторов
$M = \{m_1, \ldots, m_n\}$ набор событий.

Втунтри множества акторов событий связи отсуствуют

Представление

Матрица аффилированности $g \times n$ отражает связи каждого актора с каждым событием

$A = (a_{ij})_{1 \leq i \leq g, 1 \leq j \leq n}$, где $a_{ij} = 1$, если $i$ сов с $j$ событием

Двудольные граф -- граф, вершины которого моно разбить на две части так, что каждое ребро соединяет узлы из одного узла с другим.

$$X = \begin{pmatrix}
0 & A  \\
A^T & 0  \\
\end{pmatrix}$$

Гиперграф -- обобщение графа в котором каждым ребром могут соединнять не толька два узла но и любой подмножетво узлов.

$H = H(A, B)$

$A$ - набор вершин
$B$ - множества гипергребер.

Можно задачать матрицу совместного участия акторов $X^N = AA^T = (x_{ij}^N)_{1 \leq i, j \leq g}$

Сущеуствует $X_{kl}^M$ - количество акторов, которые участвуют в $k$ и $l$ событиях.

$$X_{kl}^M = \sum_{i = 1}^{g}a_{ik}a_{il}$$

Матрица перекрытия событий

$X^M = A^T A = (X_{kl}^M)_{1 \leq k, l \leq n}$

Свойства

1. Количество событий, с которыми связан каждый актор определены матрицей аффилированности А или матрица совместного учатстия актором $X^N$ количество событий, с которыми связаны множество акторов вычисляются

$$a_{i+} = \sum_{j = 1}^n a_{ij} = x_{ii}^N$$

2. Срденее количество событий
$$\overline{a_{i+}} = \frac{1}{g} \sum_{i = 1}^g \sum_{j = 1}^n a_{ij} = \frac{a_{++}}{g} = \frac{1}{g} \sum_{i = 1}^g x_{ij}^N$$

Количество акторос с которым связано каждое событие определяющее матрицей аффилированности А или матрицей пересечения событий $X^M$

3. Количествоа акторов, с которыми связаны события

$$a_{+j} = \sum_{i = 1}^{g}a_{ij} = x_{jj}^M$$

Среднее количество

$$\overline{a_{+j}} = \frac{1}{n}\sum_{i = 1}^g \sum_{j = 1}^na_{ij} = \frac{a_{++}}{n} =\frac{1}{n}\sum_{j = 1}^nx_{jj}^M$$

Плотность $X^M$

$$\operatorname{den}_N(G) = \frac{1}{g(g - 1)} \sum_{i,j=1,\, i \neq j}^{g}x_{ij}^N$$

$$\operatorname{den}_M(G) = \frac{1}{n(n - 1)} \sum_{k,l=1,\, l \neq k}^{n}x_{kl}^M$$

1. Длина цепи между акторами больше или равна двум.
2. Два актора достижимы через связующие их события
3. Достижимость можно исследовать с помощью социометрица  X размера $(g + n) \times (g + n)$ описывает дрвудольный граф
4. Диметр сети аффилированности это длина самой длинной цепи между всякой парой парой акторов.

Клика уровня С максимальный полный подграф из трех или более узлов, в котором
- все пары узлов участвуют не  менее чем в С событияих( для отн совместности участиноков аффилированности)
- все пыра событий имеют не менее С общих членов.


$x_{kl}^M$ - количество акторов участвующий в $k$ и $l$.

$x_{\overline{kl}}^M$ - количество акторов не участвующий в ни $k$ ни в  $l$.

$x_{\overline{k}l}^M$ - количество акторов  участвующий в  $l$ и не  в  $k$.

$x_{k\overline{l}}^M$ - количество акторов  участвующий в $k$ и не  в  $l$.

$$g = x_{kl}^M + x_{\overline{k}l}^M + x_{\overline{kl}}^M + x_{k\overline{l}}^M$$

Отношение шансов $$\Theta_{kl} = \frac{x_{kl}^M / x_{\overline{k}l}^M}{x_{k\overline{l}}^M / x_{\overline{kl}}^M}$$

Логарифм отношения шансов

$$\ln \Theta_{kl} = \ln \frac{x_{kl}^M / x_{\overline{k}l}^M}{x_{k\overline{l}}^M / x_{\overline{kl}}^M}
= \ln x_{kl}^M + \ln x_{\overline{kl}}^M - \ln x_{\overline{k}l}^M - \ln x_{k\overline{l}}^M$$
