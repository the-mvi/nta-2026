## Основные теоретические положения

**Социоматрица** - матрица X размерности g*g, описывающая узлы в сети. g - это кол-во акторов.

**Список рёбер** — это способ представления графа (сети) в виде массива, где каждая строка описывает одну существующую связь (ребро) между парой вершин.

**Алгоритм укладки** — это метод, который принимает на вход структуру сети (матрицу смежности или список рёбер) и рассчитывает геометрические координаты \((x, y)\) для каждого узла.

**Актор** — базовый элемент, представляющий отдельную единицу, участвующую во взаимодействии.

**Гиперребро** - ребро, соединящее более двух вершин.

**Мультиплексное ребро** - ребро, представляющее более 1 связи между вершинами.

**Триада** — подструктура сети, состоит из трех акторов и всех потенциальных связей между ними.
- Закрытая: есть все 3 связи.
- Открытая: есть 2 связи.

**"Пятичисловая сводка" для анализа сетей**
- Размер сети (g) - кол-во акторов сети.
- Плотность сети - доля имеющихся связей по отношению к максимально возможному кол-ву связей.
$$ den(G) = \frac{\sum_{i = 1}^g \sum_{j = 1}^g x_{ij}}{g(g-1)} $$
- Число компонент связности - подгруппа, в которой все акторы связаны друг с другом любым путём.
- Диаметр сети - максимальное расстояние между двумя узлами в сети.
Глобальный коэффициент кластеризации - мера, которая характеризхует степень "треугольности" всей сети в целом:
$$ d_t(G) = \frac{3\tau_{\triangle}(u)}{\tau_3(G)} $$

$ \tau_{\triangle}$ — количество закрытых триад.

$ \tau_3$ — количество закрытых и открытых триад.



**Центральность** — количественная мера $C_A$, определяющая относительную важность актора или его влияние в сети на основе его позиции и связей с другими акторами.

* **Центральность по степени:**
    * $C_D(n_i) = deg(n_i)$
* **Центральность по близости:**
    * $C_C(n_i) = (\sum_{j = 1}^g d(n_i, n_j))^{-1}$
* **Центральность по посредничеству:**
    * $C_B(n_i) = \sum_{i \subset K} \frac{\sigma_{ik}(n_i)}{\sigma_{ik}}$
* **Центральность по информации:**
    * $C_I(n_i) = \frac{1}{c_{ii} + (T - 2R)/g}$

**Престиж** — количественная мера $P_A$ влиятельности актора, основанная на входных связях.

* **Престиж по степени:**
    * $P_D(n_i) = deg_I(n_i)$
* **Престиж по близости:**
    * $P_P(n_i) = \frac{|I_i| / (g - 1)}{\sum_{n_j \in I_i} d(n_j, n_i) / |I_i|}$
* **Престиж по рангу:** вычисляется через решение системы уравнений
    * $p = X^Tp$, где $X$ — социоматрица.

### Задание 4

```python
basic_network <- network(s_matrix, directed = FALSE) 
# передается сама социоматрица и параметр, который берет в расчет, стоит ли считать связи направленными или нет
```

#### Аргументы функции
- x, для создания сети нужна матрица, которая задаст структуру сети; она может быть обычной, матрицей инциденции или в форме листа ребер, также можно передать другой объект класса сети.

#### Дополнительные аргументы

- vertex.attr, лист содержащий атрибуты вершин

- vertex.attrnames, лист содержащий имена атрибутов вершин

- directed, логическое значение; должны ли связи считаться направленными?

- hyper, логическое значение; могут ли существовать гиперребра?

- loops, логическое значение; могут ли быть в сети циклы?

- multiple, логическое значение; могут ли быть мультиплексные ребра?

- bipartite, двудольная ли сеть? значение параметра - кол-во акторов в одной из частей двудольной сети

- vertices, необязательный объект data.frame, содержащий атрибуты вершин.

- bipartite_col, default: "is_actor". Имя колонки, которая указывает какие вершины являются акторами в двудольной сети

- matrix.type, тип матрицы - либо "adjacency", либо "edgelist", либо "incidence". Т.е. смежности, список ребер, инциденции

- mixingmatrices, лигическое значение; должны ли мы считать объект матрицей смешивания, показывающей, как часто узлы с одинаковыми или разными атрибутами связаны

- na.omit, логическое значение; пропустить вывод сводной информации по пропущенным (NA) атрибутам сети?

- print.adj, логическое значение; вывести полную структуру матрицы смежности сети?

- object, объект класса сети, возвращаемое значение.

### Задание 5

Отобразить сеть с помощью алгоритмов укладки Фрюхтермана—
Рейнгольда и Камада—Каваи. Объяснить, в чём различие между
алгоритмами и укладками.

```python
library(sna)
par(mfrow = c(1, 2)) # дает 2 сразу в одном окне построить
plot(basic_network, 
    mode = "fruchtermanreingold", 
    main = "Фрюхтермана—Рейнгольда",
    vertex.col = "blue", 
    vertex.cex = 1.5)
plot(basic_network, 
    mode = "kamadakawai", 
    main = " Камада—Каваи",
    vertex.col = "red", 
    vertex.cex = 1.5)
```

*   **Фрюхтерман-Рейнгольд** создаёт внешне сбалансированную картинку, оптимизируя пустое пространство. Алгоритм похож на то, как отталкиваются друг от друга однозаряженные частицы.
*   **Алгоритм Камада–Каваи** задаёт целевое расстояние между каждой парой вершин, пропорциональное длине кратчайшего пути в графе, затем строит функцию «энергии системы» и минимизирует её. Таким образом, алгоритм опирается на *глобальную* структуру сети. Алгоритм похож на то, как притягиваются натянутые друг к другу пружины.

Сеть бизнесов семей Флоренции слабосвязана, в ней присутствует 5 изолированных акторов (семьи **Строцци, Ридолфи, Пуцци, Альбицци, Ацияуоли**), что не имеют связей ни с кем. При этом сама по себе сеть также не особо плотная, однако в ней наблюдается вытянутая структура с диаметром 5. Сама по себе сеть легка для анализа даже визуально. В ней присутствуют основные акторы в лице семей **Медичи и Барбадори** - они выступают в роли самых влиятельных и связующих семей в бизнесе Флоренции.

Сеть менеджеров компании Кракхардт касается того, кто к кому ходит для получения совета. Она крайне плотная, в ней высокое кол-во связей между менеджерами и при этом не наблюдается откровенных изгоев. Тут также можно выделить основных акторов: так, например, менеджер 21 является самым значимым лицом, к которому много кто ходит для совета. Наблюдается, что влиятельность и важность менеджера в сети соответствует высокому уровню доступности, т.е. показателя престижа по близости.

Заключая, можно сделать вывод, что "пятичисловая сводка" о сети дает возможность понять её общее устройство, однакое не дает конкретики об отдельных акторах в ней. А меры центральности и престижа, наоборот, позволяют выделить ключевых и не особо акторов в сети.
