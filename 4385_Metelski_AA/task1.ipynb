## Основные теоретические положения

### Определения

**Социальная сеть** - объект, представляющий собой совокупность конечного набора (или наборов) акторов и отношения (или отношений), опеределённых на них. Отношения могут бы направленными и не направленными. 

**Социоматрица** - матрица размерности g*g.

**Граф** - модель социональной сети с ненаправленными отношениями.

**Ориентированный граф** - модель социональной сети с направленными дихотомическими отношениями.

#### Алгоритмы укладки

**Алгоритм Фрюхтермана-Рейнгольда** - алгоритм моделирует граф как систему частиц (узлов), взаимодействующих через силы притяжения и силы отталкивания.

**Алгоритм Камада-Каваи** - алгоритм моделирует граф как систему пружин между всеми парами узлов, идеальная длина которых пропорциональна кратчайшему пути в графе, и ищет раскладку минимизацией суммарной энергии растяжения и сжатия этих пружин.

#### Пятичисловая сводка

1. **Размер сети** - количество акторов в сети.
2. **Плотнось сети** - доля имеющихся связей по отношению к максимально возможному количеству связей.

$$ den(G) = \frac{\sum_{i = 1}^g \sum_{j = 1}^g x_{ij}}{g(g-1)} $$

3. Число компонент связности.
4. Диаметр сети.

$$ diam(G) = \max_{n_{i}, n_{j} \in N}{d(n_i, n_j)} $$

5. Коэфициент кластеризаци:
 
    1. **Глобальный коэфициент кластеризаци** - мера, характеризующяя степень кластеризации(треугольности) всей сети в целом.

    $$cl_T(G) = \frac{3\tau_{\triangle}(G)}{\tau_3(G)}$$

    где $\tau_{△} $ - количество треугольников в котором в триаде наблюдаются ровно три связи;

    где $\tau_3$ - количетсво треугольников в которых наблюдается 2 или 3 связи.

    2. **Локальный коэфициент кластеризации** - мера, которая характеризует степень кластиеризации треугольника для конкретного узла в сети.

    $$
    cl(n_i) =
    \begin{cases}
    \frac{2N_i}{k_i(k_i - 1)}, k \leq 2 \\
    0
    \end{cases}
    $$

    где $k_i$ - число смежных с i узлов;
    
    где $N_i$ - количество ребер связаных смежных узлов между собой.

#### Центральность

- количественная мера $C_A$ определяющая относительную важность актора или его влияние в сети на основе его позиции и связей с другими акторами.

**Центральность по степени**  - мера важности узла в сети, основаная на количестве его непосредственных связей с другими узлами.

$$ C_D(n_i) = deg(n_i) $$


**Нормированная центральность** по степени - это доля от максимума возможного числа связей

$$ C_D`(n_i) = \frac{C_D(n_i)}{g - 1} $$


**Центральность по близости** - мера, определяющяя насколько узел близок ко всем остальным узлам в сети, на основе кратчайших путей.

$$C_C(n_i) = (\sum_{j = 1}^g d(n_i, n_j))^{-1}$$

**Нормированная центральность по близости**
$$C_C`(n_i) = (g - 1) C_C(n_i)$$

**Центральность по посредничеству** - мера, определяющяя насколько часто узел выступает мостом на кратчайшем путях между узлами в сети.

$$ C_B(n_i) = \sum_{i \subset K} \frac{\sigma_{ik(n_i)}} {\sigma_{ik}}$$

где $\sigma_{ij}$ - количество кротчайших путей от $n_i$ до $n_j$;

где $\sigma_{ij}(n_k)$ - количество кратчайших путей  соеди $n_i, n_j$ проходящих через $n_k$.

**Нормировання центральность по посредничеству**

$$ C_b`(n_i) = \frac{2C_b(n_i)}{(g - 1)(g - 2)}$$

**Центральность по информации** - мера важности узла в сети, основанная на его способности передавать информацию по всем возможным путям между другими узлами.

$$ C_I(n_i) = \frac{1}{c_{ii} + (T - 2R)/g}$$

**Нормированная центральность по информации**

$$C_I`(n_i) = \frac{C_I(n_i)}{\sum_{i = 1}^{g} C_I(n_j)}$$


#### Престиж

- количественная мероа $P_A$ влиятельности актора основанная на входящих связях.

  
**Престиж по степени** - мера статуса или влиятельности узла  в оргграфе оснащенной на количестве входный связей.

 $$P_D(n_i) = deg_I(n_i)$$

**Нормированный престиж по степени**

 $$ P`_D(n_i) = \frac{P_D(n_i)}{g - 1} $$

**Престиж по близости** - мера статуса узла в орграфе, учитывая долю узлов, достажимости для актора и среды $\rho$.

 $$ P_P(n_i) = \frac{|I_i| / (g - 1)}{\sum_{n_j \in I_i} d(n_j, n_i) / |I_i|}$$

 где $I_i$ - множество актороов достижимых для $n_i$.

**Престиж по рангу** - мера влиятельности узла в направленной сети, учитывает не только количество входных связей, но и "авторитетность" тех,  кто эти связи проставляет.

  - Ранг актора $n_i$ является ф-ей ранга авторов, которые быбирают $n_i$ $P_R(n_i) = x_1  P_R(n_1) + ... + x_{g_i}P_R(n_g)$
  - Имеется g линейных уравнений с g неизвестными. Пусть $p = (P_R(n_1), ... , P_R(n_g))^T$. Тогда СЛУ будет иметь вид

  $$p = X^Tp$$

или

  $$(E - X^T)p = 0$$

Вводим параметр $a > 0$ - параметр затухания.
- x - вектор столбец суммы столбов социоматрицы Х.
- Вектор рангов p находится из СУ
$$(\frac{1}{a}E - X^T)p = x$$

  - параметр а выбирается самостоятельнт исходя из условий

$$ \lambda_1 < \frac{1}{a} < 2 \lambda_1$$

где $\lambda_1$ - наибольшое веществуенное собственное число социоматрицы X.

**Нормированный престиж по рангу**
$$ P_R`(n_i) = \frac{P_R(n_i)}{m}$$

где $m = \sum_{k = 1}^{g - 1} a^k A_{g - 1}^k$.

## Выполнение работы

```r
library(igraph) 
```

```r
data <- read.csv('florentine_marital.csv', header = FALSE, comment.char = '#')
# считали данные с диска, комменты скипаем

names <- as.character(unlist(data[1, ]))
# Получили вектор имён

matrix <- as.matrix(data[-1, ])
#преобразуем в матрицу все данные, за исключением первой строки с названиями, на этом этапе имеем матрицу строк
mode(matrix) <- 'numeric'
#меняем тип данных на численный
```

```r
edge_list <- which(matrix == 1 & row(matrix) <= col(matrix), arr.ind = TRUE)
#получили список координат 1ц в матрице, т.к. граф не ориентированный, то смотрим верхний треугольник
#arr.ind = TRUE - нужна чтобы возвращался не номер в векторе, а коориданат из номера строки и столбца
print(edge_list[1:5, ])
#п.2. вывели первые 5 элементов списка ребер
```

```r
rownames(matrix) <- names
colnames(matrix) <- names
#п.3. заименуем строки и столбцы матрицы
print(matrix)
```

```r
web <- graph_from_adjacency_matrix(matrix, mode = 'undirected', weighted = NULL)
#п.4 строим направленную сеть. mode = 'undirected' - задаёт что граф двунаправленный
# weighted = NULL - граф не взвешенный
print(web)
```

```r
#п.5. отобразили сеть с помощью алгоритмов укладки Фрюхтермана—Рейнгольда и Камада—Каваи
# укладка Фрюхтермана-Рейнгольда
plot(web, 
     layout = layout_with_fr(web), 
     main = 'Фрюхтерман-Рейнгольд',
     vertex.color = 'skyblue',
     vertex.size = 25,
     vertex.label.color = 'black')

# укладка Камада-Каваи
plot(web, 
     layout = layout_with_kk(web), 
     main = 'Камада-Каваи',
     vertex.color = 'skyblue', 
     vertex.size = 25,
     vertex.label.color = 'black')
#layout_with_kk/layout_with_fr функции из igraph которые и делают укладку
```

Объяснить, в чём различие между алгоритмами и укладками:
1. Ну минимум у Фрюхтермана-Рейнгольда ребра не пересекаются, т.е. нарисованный граф плоский. (в ориентированном графе начали пересекаться, не знаю завсисит ли от этого)
2. В Фрюхтермана-Рейнгольда считаем силы отталкивания в Камада-Каваи энергию. На одной иттерации в первом перераспределяем все узлы, а в втором меняем только один узел.
3. Насколько говорит интернет Камада-Каваи в большей мере показывает реальное положение вершин друг относительно друга в графе.

```r
summary(web)
#расшифруем что она вывела 196986c(хэш) U(ненаправленный)N(есть аттрибут имени)-- 16(кол-во вершин) 20(кол-во ребер) --
#п.6 вывел описательную сводку
```

```r
cat(vcount(web), "/", ecount(web), "\n")
#п.7.a. размер сети. Кол-во вершин/кол-во ребер

density_hand <- 2 * ecount(web) / (vcount(web) * (vcount(web) - 1))
density_auto <- edge_density(web)
cat(density_hand, "/", density_auto, "\n")
#п.7.b. плотность сети. Формула для ненаправленного графа 2L/(g)(g-1), получается из L/C(2, g). Граф не плотный

print(components(web)$no)
#п.7.c. всего 2 компоненты связности, проверяем функцией components, $no 
#позволяет вывести только колиество. При неплотном графе всего 2 комп связности

print(diameter(web, directed = FALSE))
#п.7.d. вычисляем диаметр, directed - говорим ориентированный граф или нет. диаметр = 5. При 16 и 20 ребрах 
#вершинах получаем достаточно маленький диаметр, несмотря на низкую плотность

cat("global: ", transitivity(web, type = 'global'), "\n")
#кластеризация или треугольность графа достаточно низкая 
cat("local: ", transitivity(web, type = 'local'), "\n")
#у вершин со степенью меньше 2х, такой коээфициент = NaN. Локальный коэффициент кластеризации показывает
#долю замкнутых треугольников вокруг конкретной вершины. Среди множества вершин достаточно вершин с единственной связью. 
#Также нет ни одной вершины которую окружают только треугольники. 
#п.7.c. Вычислил коээфициент кластеризации
```

```r
#п.8. Посчитаем различные меры центральности
webm <- induced_subgraph(web, which(degree(web) > 0))
#induced_subgraph забирает из переданного графа только те вершины и их ребра для которых выолняется условие. 

cat(degree(webm, normalized = TRUE))
#нормированная центральность по степени
print(closeness(webm, normalized = TRUE))
#нормированная центральность по близости
print(betweenness(webm, normalized = TRUE))
#нормированая центральность по посредничеству

to_matrix <- as_adjacency_matrix(webm, sparse = FALSE)
#sparse = FALSE - для того чтобы создалась классическая матрица а не какая-то с ухищрениями для экономии памяти
print(sna::infocent(to_matrix, gmode = 'graph', rescale = TRUE))
#rescale = TRUE - для нормировки; gmode = 'graph' - не ор. граф
#нормированная цетральность по информации

cat("\n\n")
#соберём всё в одну таблицу
table <- data.frame(
    CD = degree(webm, normalized = TRUE),
    CC = closeness(webm, normalized = TRUE),
    CB = betweenness(webm, normalized = TRUE),
    CI = sna::infocent(to_matrix, gmode = 'graph', rescale = TRUE)
)
print(round(table, 3))
#ограничимся тремя числами после запятой
```
