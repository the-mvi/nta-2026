## Основные теоретические положения

### Определения

**Точки сочленения** - узел, при удалении которого увеличивается число компонент связности.

**Мосты** -  ребро, при удалении которого увеличивается число компонент связности.

#### Мера централизации

**Централизация группы** - количественная мера $C_A$ определяющая относильную изменчивость или неоднородность центральности акторов в группе.

Общая формула:
$$ C_A=\frac{\sum_{i=1}^g (C_A(n^*)-C_A(n_i))}{max(\sum_{i=1}^g (C_A(n^*)-C_A(n_i)))} $$
где $C_A(n^*)=\max_i(C_A(n_i))$

**Централизация по степени**
$$ C_D=\frac{\sum_{i=1}^g (C_D(n^*)-C_D(n_i))}{(g-1)(g-2)} $$

**Централизация по близости**
$$ C_C=\frac{\sum_{i=1}^g (C_C'(n^*)-C_C'(n_i))}{\frac{(g-2)(g-1)}{2g-3}}  $$

**Централизация по посредничеству**
$$ C_B=\frac{1}{g-1}\sum_{i=1}^g(C_B'(n^*)-C_B'(n_i)) $$

**k-ядро** - максимальный индуцированный подграф, в котором каждая вершина имеет степень не меньше k внутри этого подграфа. Его можно получить последовательным удалением вершин, степень которых меньше k, пока таких вершин не останется.

**Число ядра** вершины — наибольшее k, для которого она входит в k-ядро. Вершины с большим числом ядра обычно принадлежат к более плотной центральной части сети.

#### Сети аффелированносоти
**Сети аффелированносоти** - сеть, в которой участники аффелированы на основе сотрудничества в рамках какой-то группы или совместного участия в опред. событии. 
- Связи между акторами основываются на связи учасия в группе или событии

$N=\{n_i, ... n_g\}$  - набор акторов

$M=\{m_1, ...m _n\}$ - набор событий

$n_i\in N связаны с m_j \in M$, при этом связи внутри M и N отсутсвуют.

**Представление сетей аффелированности**

- Матрица аффелированности
$$A=(a_{ij})_{g*h}$$
$$a_{ij}=
\begin{cases}
1 \text{, если i-актор связан с j-ым событием} \\
0 \text{, иначе}
\end{cases}

- Двудольный граф
$$
\begin{pmatrix}  % круглые скобки ( )
0_{s*s} & A \\
A^T & 0_{h*h}
\end{pmatrix}
$$

- Гиперграф
обощение графа, в котором каждым ребром могут соединяться не только два узла, но и любые подмножества узлов.
$$H=H(A, B)$$
$$A=\{a_1, ... a_g\}$$
$$B=\{b_1, ... b_n\}$$


##### СА как одномодальные сети
Пусть $x_{ij}^N$ - кол-во событий, в которых учавствует i и j актора

$x_{ij}^N=\sum_{k=1}^na_{ik}a_{jk}$
где k - события

$X^N=AA^T=(x_{ij}^N)_{g*g}$ - матрица совместного участия

Пусть $x_{kl}^M$ - количество акторов, которые учавствуют в k и l событиях

$x_{kl}^M=\sum_{i=1}^ga_{ik}a_{il}$

$X^M=A^TA=(x_{kl}^M)_{h*h}$ - матрица совместного участия

##### Свойства

Количество событий, которыми связан каждый актор:
$$a_{ik}=\sum_{j=1}^na_{ij}=x_{ii}^N$$

- Среднее количество событий
$$\overline{a_{i+}}=\frac{1}{g}\sum_{i=1}^gn_{ii}^N$$

- Среднее количество акторов
$$\overline{a_{+j}}=\frac{1}{h}\sum_{i=1}^hn_{jj}^M$$

- Количество акторов, связанных с событием
$$a_{+j}=\sum_{i=1}^ga_{ij}=x_{jj}^M$$

- Плотность $x^N$
$$den_N(G)=\frac{1}{g(g-1)}\sum_{i,j=1; i\neq j}^hx_{ij}^N$$

- Плотность $x^M$
$$den_M(G)=\frac{1}{h(h-1)}\sum_{k,l=1; k\neq l}^hx_{kl}^M$$

- Диаметр
длина самой длиной цепи между любой парой акторов и/или событий

##### Меры
$x_{kl}^M$ - кол-во участников k и l

$x_{k\overline{l}}^M$ - кол-во участников k, но не l

$x_{\overline{k}l}^M$ - кол-во участников l, но не k

$x_{\overline{kl}}^M$ - кол-во участников не k и l

(ну я хз как это в техе записать, поэтому будет так криво)

Отношение шансов:
$$\theta=\frac{x_{ll}^M - x_{\text{не}k\text{ не}l^M}}{x_{\text{не}kl}^M-x_{k\text{ не}l}^M}$$

Логарифм отношения шансов:
$$ ln\theta_{kl} = \ln{x_{kl}^M} + \ln{x_{\text{не}k\text{ не}l}^M} - \ln{x_{\text{не}kl}^M} - \ln{x_{k\text{ не}l}^M} $$

```r
library(igraph) 
library(sna)
```

```r
data <- read.csv('florentine_marital.csv', header = FALSE, comment.char = '#')
# считали данные с диска, комменты скипаем

names <- as.character(unlist(data[1, ]))
# Получили вектор имён

matrix <- as.matrix(data[-1, ])
#преобразуем в матрицу все данные, за исключением первой строки с названиями, на этом этапе имеем матрицу строк
mode(matrix) <- 'numeric'
#меняем тип данных на численный
rownames(matrix) <- names
colnames(matrix) <- names
#заименуем строки и столбу
print(matrix)
```

```r
web <- graph_from_adjacency_matrix(matrix, mode = 'undirected', weighted = NULL)
#п.1 строим ненаправленную сеть. mode = 'undirected' - задаёт что граф двунаправленный
# weighted = NULL - граф не взвешенный
plot(web, 
     layout = layout_with_fr(web), 
     main = 'Фрюхтерман-Рейнгольд',
     vertex.color = 'skyblue',
     vertex.size = 25,
     vertex.label.color = 'black')
#вывеои граф
```

```r
#п.2.
ap <- articulation_points(web)
cat("Точки сочленения\n")
print(ap)
b <- bridges(web)
cat("Мосты\n")
print(b)

v_col <- rep('skyblue', vcount(web))
#(rep - повторить)
v_col[ap] <- 'red'
e_col <- rep('grey', ecount(web))
e_col[b] <- 'green'
plot(web, layout = layout_with_fr(web), vertex.color = v_col, edge.color = e_col,
     main = 'Точки сочленения и мосты')
```

```r
# п.3
#Убираем изолированные вершины
webm <- induced_subgraph(web, which(igraph::degree(web) > 0))
A <- as.matrix(as_adjacency_matrix(webm, sparse = FALSE))

#для наглядности нарисуем
plot(webm, 
     layout = layout_with_fr(webm), 
     main = 'Фрюхтерман-Рейнгольд',
     vertex.color = 'skyblue',
     vertex.size = 25,
     vertex.label.color = 'black')
```

```r
#тоже п. 3
# Централизация по четырём мерам
cat('По степени:\n')
print(centr_degree(webm, loops = FALSE))
cat('\nПо близости:\n') 
print(centr_clo(webm, mode = 'all'))
cat('\nПо посредничеству:\n')
print(centr_betw(webm, directed = FALSE))
cat('\nПо информации:\n') 
print(sna::centralization(A, sna::infocent, mode = 'graph', rescale = TRUE))
```

Итого:
- По степени: 0.2747253
- По близости: 0.3224523
- По посредничеству: 0.4368132
- По информации: 0.065451

Наиболее сильная централизация по посредничеству, т.е. распределение кратчайших путей сильно зависит от небольшого числа вершин

```r
#п.4
core <- coreness(web)
print(core)
#Всего 2 ядра, первое содержит все кроме Pucci, 
#Второе содержит: Albizzi, Barbadori, Bischeri, Castellani, Guadagni, Medici, Peruzzi, Ridolfi, Strozzi, Tornabuoni 
#Максимальное число ядра равно 2, следовательно, в сети отсутствует 3-ядра
```

```r
#п.5
#Для 6-того варианта: InfoMAP, Louvain, Optimal (вроде бы....)

community_1 <- cluster_infomap(webm)
plot(webm, 
     layout = layout_with_fr(webm), 
     vertex.color = membership(community_1), 
     main = 'InfoMAP'
    )

community_2 <- cluster_louvain(webm)
plot(webm, 
     layout = layout_with_fr(webm), 
     vertex.color = membership(community_2),
     main = 'Louvain')

community_3 <- cluster_optimal(webm)
plot(webm, 
     layout = layout_with_fr(webm), 
     vertex.color = membership(community_3),
     main = 'Optimal')

cat('InfoMAP, модульность:', modularity(community_1), '\n')
cat('Louvain, модульность:', modularity(community_2), '\n')
cat('Optimal, модульность:', modularity(community_3), '\n')
```

Модулярность показывает, насколько больше связей находится внутри найденных сообществ по сравнению со случайной моделью с теми же степенями вершин. Чем выше значение, тем более выражено разделение на сообщества. Наилучшее значение модульности получено алгоритмами Louvain и Optimal. При этом оба алгоритма сформировали одинаковое разбиение сети на три сообщества.

**Описание:**
- InfoMAP - моделирует случайное блуждание по сети, этим оценивает потоки информации между узлами
- Louvain - жадный алгоритм, стиримится к максимуму модулярности
- Optimal - алгоритм не использующий эвристики, находит гарантировано оптимальное разбиение с точки зрения максимизации модульности

```r
#п.14
pairs <- combn(1:ncol(ceo_matrix), 2)
log_OR <- rep(NA, ncol(pairs))

for (i in 1:ncol(pairs)) {
  x <- ceo_matrix[, pairs[1, i]]
  y <- ceo_matrix[, pairs[2, i]]
  n11 <- sum(x == 1 & y == 1)
  n10 <- sum(x == 1 & y == 0)
  n01 <- sum(x == 0 & y == 1)
  n00 <- sum(x == 0 & y == 0)
  log_OR[i] <- log(((n11 + 0.5) * (n00 + 0.5)) / ((n10 + 0.5) * (n01 + 0.5)))
}

odds_table <- data.frame(Event_1 = colnames(ceo_matrix)[pairs[1, ]],
                         Event_2 = colnames(ceo_matrix)[pairs[2, ]], log_OR)
print(odds_table[order(odds_table$log_OR, decreasing = TRUE), ])
```

Для двух событий логарифм отношения шансов > 0 означает, что участие в одном событии связано с повышенными шансами участия во втором. 

Значение < 0 означает обратное. 

Значение около нуля соответствует слабой зависимости.
