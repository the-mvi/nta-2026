## Основные теоретические положения

### Определения

**Социальная сеть** - совокупность конечного набора акторов и отношений между ними. Отношения могут быть направленными и ненаправленными.

**Социоматрица** - квадратная матрица размерности `g × g`, описывающая отношения между акторами сети.

**Граф** - модель социальной сети с ненаправленными отношениями.

**Ориентированный граф** - модель социальной сети с направленными дихотомическими отношениями.

### Алгоритмы укладки

**Алгоритм Фрюхтермана - Рейнгольда** - алгоритм размещения вершин графа, который моделирует их как частицы, взаимодействующие посредством сил притяжения и отталкивания.

**Алгоритм Камада - Каваи** - алгоритм размещения вершин, основанный на модели пружин между парами узлов. Идеальная длина каждой пружины зависит от кратчайшего пути между соответствующими вершинами. Алгоритм ищет расположение с минимальной суммарной энергией системы.

### Пятичисловая сводка

1. Размер сети (англ. network size) - количество акторов в сети.

2. Плотнось сети (англ. network density) - доля имеющихся связей по отношению к максимально возможному количеству связей.

$$
den(G) = \frac{\sum_{i=1}^{g}\sum_{j=1}^{g}x_{ij}}{g(g-1)}
$$


3. Компонента связности (англ. component) - подгруппа, в которой все акторы связаны друг с другом.

4. Диаметр сети (англ. network diameter) - это максимальное расстояние между двумя узлами.

$$
diam(G) = \max_{n_i,n_j \in N} d(n_i,n_j)
$$

5. Глобальный коэфициент кластеризаци (англ. global clustering coefficient) - мера, кооторая характеризует степень кластеризации( или "треугольности") всей сети в целом.

$$d_t(G) = \frac{3\tau_{\triangle}(u)}{\tau_3(G)}$$

где $\tau_{△} $ - количество закрытых треугольников (триад, где наблюдаются ровно три связи);

где $\tau_3$ - количество открытых и закрытых треугольников.


**Локальный коэфициент кластеризации**  (англ. local clustering coefficient) - мера, которая характеризует степень кластиеризации треугольника (или "треугольности") для конкретного узла в сети.


$$
    cl(n_i) =
    \begin{cases}
    \frac{2N_i}{k_i(k_i - 1)}, k \leq 2 \\
    0
    \end{cases}
$$

$k_i$ - число смежных узлов с i-ым узлом;

$N_i$ - количество ребер, связывающие смежные с i-ым узлом между собой.


### Центральность

**Центральность** (англ. centrality) - количественная мера $C_A$ определяет относительную важность актора или его влияюние в сети на основе его позиции и связей с другими акторами.

**Центральность по степени** (англ. degree centrality) - мера важности узла в сети, основанная на количестве его непосредственных связей с другими узлами:

$$
C_D(n_i) = deg(n_i)
$$

**Нормированная центральность по степени** показывает долю существующих связей от максимально возможного количества.

$$ C_D(n_i)` = \frac{deg(n_i)}{g - 1} = \frac{C_D(n_i)}{g - 1} $$

**Центральность по близости** (англ. closeness centrality) - мера, определяющяя насколько узел близок ко всем остальным узлам в сети на основе кратчайших путей.

$$
C_C(n_i) = \left(\sum_{j=1}^{g}d(n_i,n_j)\right)^{-1}
$$

**Нормированная центральность по близости**:

$$
C_C'(n_i) = (g-1)C_C(n_i)
$$

**Центральность по посредничеству** показывает, насколько часто вершина участвует в кратчайших путях между другими вершинами сети.

$$
C_B(n_i) = \sum_{s \neq i \neq t}\frac{\sigma_{st}(n_i)}{\sigma_{st}}
$$

Здесь $\sigma_{st}$ - количество кратчайших путей между вершинами $n_s$ и $n_t$, а $\sigma_{st}(n_i)$ - количество таких путей, проходящих через вершину $n_i$.

**Нормированная центральность по посредничеству**:

$$
C_B'(n_i) = \frac{2C_B(n_i)}{(g-1)(g-2)}
$$

**Центральность по информации**(англ. information centrality) - мера важности узла в сети, основанная на его способности передавать информацию по всем возможным путям между другими узлами.

$$
C_I(n_i) = \frac{1}{c_{ii} + (T-2R)/g}
$$

**Нормированная центральность по информации**:

$$
C_I'(n_i) = \frac{C_I(n_i)}{\sum_{j=1}^{g}C_I(n_j)}
$$

### Престиж

 **Престиж** (англ. prestige) - количестваенная мера $P_A$ влиятельности актора основанная на входящих связях.

 **Престиж по степени** (англ. degree prestige) - мера статуса или влиятельности узла  в орграфе, оснащенная на количестве входный связей.

$$
P_D(n_i) = deg_I(n_i)
$$

**Нормированный престиж по степени**:

$$
P_D'(n_i) = \frac{P_D(n_i)}{g-1}
$$

 **Престиж по близости** (англ. proximity prestige) - мера статуса узла в орграфе, учитывающая долю узлов, достижимых для актора и среднее расстояние до этих узлов:

$$
P_P(n_i) = \frac{|I_i|/(g-1)}{\sum_{n_j \in I_i}d(n_j,n_i)/|I_i|}
$$

Где $I_i$ - множество акторов, достижимых для актора $n_i$.

**Престиж по рангу** учитывает не только количество входящих связей, но и авторитетность акторов, которые направляют эти связи.

Ранг актора $n_i$ определяется рангами акторов, которые выбирают его:

$$
P_R(n_i) = x_1P_R(n_1) + \ldots + x_gP_R(n_g)
$$

Пусть вектор рангов имеет вид:

$$
p = (P_R(n_1), \ldots, P_R(n_g))^T
$$

Тогда система линейных уравнений записывается следующим образом:

$$
p = X^Tp
$$

или

$$
(E-X^T)p = 0
$$

Вводится параметр затухания $a > 0$. Вектор $x$ - столбец сумм столбцов социоматрицы $X$. С учётом параметра затухания вектор рангов определяется из системы:

$$
\left(\frac{1}{a}E-X^T\right)p = x
$$

Параметр $a$ выбирается из условия:

$$
\lambda_1 < \frac{1}{a} < 2\lambda_1
$$

Здесь $\lambda_1$ - наибольшее вещественное собственное число матрицы $X$.

**Нормированный престиж по рангу**:

$$
P_R'(n_i) = \frac{P_R(n_i)}{m}
$$

где

$$
m = \sum_{k=1}^{g-1}a^kA_{g-1}^k
$$

```r
# Пункт 1
# Чтение матрицы смежности
courtesy <- read.csv(
  "florentine_business.csv",
  header = FALSE, # первая строка не заголовок
  skip = 2, # скипнем две первые строки где нет данных, а имена и тд
  check.names = FALSE # не меняем данные под R, сами потом сделаем
)


# Берем первые две строки данных
court <- strsplit(
    readLines("florentine_business.csv", n = 2)[2], # n = 2 тут у нас берутся первые две строки, но чтобы была только с фамилиями - [2]
    ",\\s*"
)[[1]] # сплит бьет данные в список и в первой ячейке у нас лежит наши фамилии

# проверим сколько семей
length(court)

# Напечатаем Матрицу смежности
courtesy_ <- as.matrix(courtesy)
# Числовой тип
storage.mode(courtesy_) <- "numeric"


# Сделаем проверку
dim(courtesy_)
courtesy_
```

```r
# чтобы взять ребра между вершинами, так как у нас здесь неориентированный граф мы должны взять только верхне или нижнетреугольную матрицу.

edge <- which(
  courtesy_ == 1 & upper.tri(courtesy_),
  arr.ind = TRUE # храним координаты x и y - строк и столбцов
)
edge
```

```r
# Пункт 4
# Построим ненаправленную сеть
# graph_from_adjacency_matrix() - создание графа из социоматрицы
courage = graph_from_adjacency_matrix(courtesy_, mode = "undirected", diag = FALSE) # mode = "undirected" - чтобы граф был неориентированным и diag = FALSE, чтобы не было петель из главной диагонали
courage
```

```r
# Алгоритм укладки Фрюхтермана—Рейнгольда
# Пункт 5

plot(
  courage,
  layout = layout_with_fr(courage),
  vertex.size = 23,
  main = "Фрюхтерман-Рейнгольд" # заголовок
)
```

```r
# Алгоритм укладки Камада—Каваи

plot(
  courage,
  layout = layout_with_kk(courage),
  vertex.label = V(courage)$name,
  vertex.size = 23,
  vertex.label.cex = 1,
  main = "Камада-Каваи"
)
```

**Различие между алгоритмами Фрюхтермана - Рейнгольда и Камада — Каваи**

1. Алгоритм Фрюхтермана - Рейнгольда основан на модели физических сил: вершины отталкиваются друг от друга, а рёбра как бы притягивают связанные вершины. В алгоритме Камада - Каваи расположение вершин определяется минимизацией энергии, которая зависит от расстояний между вершинами в графе и на рисунке.

2. Алгоритмы отличаются и способом оптимизации. В Фрюхтермана - Рейнгольда на каждой итерации обновляется расположение всех вершин, а в Камада - Каваи оптимизация обычно выполняется последовательно для отдельных вершин с целью уменьшить энергию системы.

3. В результате укладки могут выглядеть по-разному. Фрюхтерман - Рейнгольд обычно хорошо показывает общую структуру и группы связанных вершин. Камада - Каваи старается отразить расстояния между вершинами: близкие по графу вершины располагаются ближе, а далёкие - дальше. Однако ни одна из укладок не гарантирует отсутствия пересечений рёбер или точного отображения какого-то единственного "реального" расположения графа.

```r
#Пункт 8. Центральность вершин без изолированных

core <- induced_subgraph(courage, V(courage)[degree(courage) > 0])
A <- as.matrix(as_adjacency_matrix(core, sparse = FALSE))

centrality <- data.frame(
  ni = V(core)$name,
  CD = degree(core) / (vcount(core) - 1),
  CC = closeness(core, normalized = TRUE),
  CB = betweenness(core, directed = FALSE, normalized = TRUE),
  CI = as.numeric(sna::infocent(A, gmode = "graph", rescale = TRUE))
)

centrality[, -1] <- lapply(centrality[, -1], round, 3)
print(centrality)

# CD - количество связей вершины относительно остальных

# CC - насколько близко вершина расположена к другим
# CB - насколько часто вершина выступает посредником
# CI - информационная центральность
```

По таблице видно, что наибольшая степень центральности у Medici (0.5), то есть у этого семейства больше всего связей. По близости к остальным лидируют Barbadori (0.588), Medici и Peruzzi (по 0.526). Посредниками чаще всего выступают Barbadori (0.556) и Medici (0.533). По показателю CI наибольшее значение у Barbadori (0.124). Изолированные вершины в таблицу не включались.

```r
# Пункт 9.

advice_matrix <- as.matrix(read.csv(
  "krackhardt_advice.csv",
  header = FALSE,
  comment.char = "#",
  check.names = FALSE
))

storage.mode(advice_matrix) <- "numeric"

# Проверим, что матрица квадратная
dim(advice_matrix)
stopifnot(nrow(advice_matrix) == ncol(advice_matrix))

# Названия вершин, так как в файле их нет
advice_names <- paste0("V", seq_len(nrow(advice_matrix)))
rownames(advice_matrix) <- advice_names
colnames(advice_matrix) <- advice_names

advice_matrix

# Список дуг. В отличие от прошлого графа берем все единицы,
# потому что здесь связи направленные
arc_pos <- which(advice_matrix == 1, arr.ind = TRUE)

advice_edges <- data.frame(
  from = rownames(advice_matrix)[arc_pos[, 1]],
  to = colnames(advice_matrix)[arc_pos[, 2]]
)

head(advice_edges, 5)

# Строим направленный граф
advice <- igraph::graph_from_adjacency_matrix(
  advice_matrix,
  mode = "directed",
  diag = FALSE
)

# Рисуем граф двумя способами
plot(advice, layout = igraph::layout_with_fr(advice),
     vertex.size = 18, vertex.label.cex = 0.7,
     main = "Фрюхтерман-Рейнгольд")

plot(advice, layout = igraph::layout_with_kk(advice),
     vertex.size = 18, vertex.label.cex = 0.7,
     main = "Камада-Каваи")

# Общая информация о графе
summary(advice)

# 9a.
n <- igraph::vcount(advice)
m <- igraph::ecount(advice)

cat("Вершин:", n, "\n")
cat("Дуг:", m, "\n")

# 9b.
# Для орграфа максимально возможно n*(n-1) дуг
D1 <- m / (n * (n - 1))
D2 <- igraph::edge_density(advice, loops = FALSE)

cat("Плотность вручную:", D1, "\n")
cat("Плотность через igraph:", D2, "\n")

# 9c.
# Слабая - не учитываем направление дуг
# Сильная - учитываем направление дуг
comp_weak <- igraph::components(advice, mode = "weak")
comp_strong <- igraph::components(advice, mode = "strong")
# Сильная компонента: между вершинами существуют пути в обоих направлениях
cat("Слабых компонент:", comp_weak$no, "\n")
cat("Сильных компонент:", comp_strong$no, "\n")

# 9d.
d <- igraph::distances(advice, mode = "out")
d <- d[is.finite(d) & row(d) != col(d)]

if (length(d) > 0) {
  print(max(d))
} else {
  print(NA)
}

# 9e.
# Здесь направление дуг не учитываем
advice_undirected <- igraph::as_undirected(advice, mode = "collapse")

cat("global: ",
    igraph::transitivity(advice_undirected, type = "global"), "\n")
cat("local: ",
    igraph::transitivity(advice_undirected, type = "local"), "\n")
```

```r
# Пункт 10

# Престиж по входящим связям
PD <- as.numeric(sna::prestige(
  advice_matrix, gmode = "digraph", cmode = "indegree"
)) / (n - 1)

# Считаем расстояния по направлению дуг
d <- igraph::distances(advice, mode = "out")
d[!is.finite(d)] <- NA
diag(d) <- NA

# Сколько вершин могут добраться до каждой вершины
reachable <- colSums(!is.na(d))

# Среднее расстояние до каждой вершины
mean_dist <- colSums(d, na.rm = TRUE) / pmax(reachable, 1)

# Престиж по расстояниям
PP <- (reachable / (n - 1)) / mean_dist
PP[reachable == 0] <- 0

# Престиж по собственному вектору
PR <- as.numeric(sna::prestige(
  advice_matrix, gmode = "digraph", cmode = "eigenvector"
))
PR <- PR / sum(PR)

# Собираем таблицу
prestige <- data.frame(
  ni = advice_names,
  PD = PD,
  PP = PP,
  PR = PR
)

prestige[, -1] <- lapply(prestige[, -1], round, 3)
print(prestige)
# PD — престиж по числу входящих связей
# PP — престиж по расстояниям до вершины
# PR — престиж по собственному вектору
```

По таблице видно, что по PD лидируют V2 (0.90), V18 и V21 (по 0.75). По PP также лидирует V2 (0.909), а по PR — V2 (0.097). Значит, V2 занимает наиболее значимое положение в сети по всем трём показателям.
