## Основные теоретические положения

Социоматрица - это матрица $\mathbb X$ размера $g\times{g}$, описывающая смежные узлы в сети $$\mathbb X = (x_{ij})_{g\times g}$$
Социальная сеть - это объект, представляющий собой совокупность конечного набора акторов и отношений, определённых на них.

Ненаправленная сеть - это модель социальной сети с ненаправленным дихотомическим отношением

Алгоритм укладки - это способ отображения графа в выводе таким образом, чтобы человеческому восприятию было проще опознать существующие в нём закономерности

А.У. фрюхтермана-рейнгольда

А.У. Камада-Каваи

Разлицие между А.У.

Описательная сводка сети - это

Пятичислова сводка сети - это набор из следующих числовых характеристик:

Размер сети - это количество акторов сети

Плотность сети - это доля связей в сети от максимально возможного числа связей
$$den(G) = \frac{\sum_{i}\sum_{j}x_{ij}}{g(g -1)} = \frac{l}{C^2_g}$$

Компонента в сети - это подгруппа, все акторы которой связаны прямо или косвенно

Диаметр сети - это максимальное растояние между двумя узлами в сети
  $$diam(G) = max_{n_i, n_j \in N} d(n_i, n_j)$$

Коэффициент кластеризации - мера, характеризующая степень треугольности всей сети   $$cl_t(G) = \frac{3T_{\Delta}(G)}{T_3(G)}$$
 где
 $T_{\Delta}(G)$ - число триад с 3 связями,
 $T_3(G)$ - число триад с 2 или 3 связями

Центральность (мера центральности) - это количественная мера $C_A$ об относительной важности актора или влиянии в сети на основе позиции и связей с другими акторами. Бывает следующих видов: по степени, близости посредничеству, информации. Для ненаправленных сетей:
$C^`_D(n_i)$ , $C^`_C(n_i)$, $C^`_B(n_i)$, $C^`_I(n_i)$:
$$C^`_D(n_i) = \frac{deg(n_i)}{g-1} = \frac{C_D(n_i)}{g-1}\\\\(g - \text{размер сети = кол-во акторов)} \\ \text{[нормир. центральность по степени]}$$
  $$C'_C (n_i) = \frac{g-1}{\sum_{j=1}^g d(n_i, n_j)}\\\\ \text{ [нормир. центральность по близости]}$$
$$C'_B (n_i) = \frac{C_B(n_i)}{C^2_{g-1}}\\ \text{где }C_B(n_i) = \sum_{j<k} \frac{σ_{jk}(n_i)}{σ_{jk}}\\σ_{jk} \text{ - количество кратчайших путей из }n_j \text{ в } n_k,\\ σ_{jk}(n_i)\text{  - количество кратчайших путей из }n_j\text{ в }n_k, \text{ проходящих через }n_i \\ \text{ [нормир. центральность по посредничеству]}$$
  $$C'_I (n_i) = \frac{C_I(n_i)}{\sum_{i=1}^g C_I(n_i)}\\C_I(n_i) = \frac{1}{c_{ii} + \frac{T-2R}{g}}\\ \text{где }A = (\mathbb{1} + K) \space - \text{матрица псевдосмежности,}\\ K\space-\text{матрица Кирхгофа,}\\ C=A ^{-1},\space T=tr(C),\space R=\sum_{j=1}^g c_{ij}\\ \text{ [нормир. центральность по информации] }$$
Мера престижа - количетвенная мера $P_A$ влиятельности актора, оcнованная на входящих связях. Бывает следующих видов:

$$P'_D(n_i)=\frac{deg_{in}(n_i)}{g-1}\\\\ \text{ [нормир. престиж по степени]}$$

$$P_P(n_i)=\frac{|I_i|/(g-1)}{\sum_{n_j \in I_i} d(n_j, n_i)/|I_i|}\\\text{где }I_i -\text{ множество акторов, достижимых в }n_i\\ \text{ [престиж по близости]}$$
  $$P'_R(n_i) = \frac{P_R(n_i)}{\sum_{k=1}^{g-1} a^k \cdot A_{g-1}^k}\\P_R(n_i) = x_{1i}  \cdot  P_R(n_1) + x_{2i}  \cdot  P_R(n_2) + ... + x_{gi} \cdot P_R(n_g)\\a-\text{ параметр затухания}\\ \text{ [престиж по рангу]}$$

Алгоритм FR:
  1) Пусть рёбра стягивают узлы друг к другу, а те наоборот стремятся удалиться друг от друга подальше.
  2) Высчитав итеративно итоговое состояние системы, получим укладку.

Итеративность приводит к тому, что нельзя гарантировать, что выходное изображение будет каждый раз одинаковым.

Условие выхода вида:

$$|i_{n+1}-i_n| \rightarrow result,\\ result\leq const \\\text{ - где }const\text{ - критерий остановки итерирования по причине достижения нужной точности.}$$

Тут высчитывается расстояние между узлами и компонентами связности на каждом шаге, пока не станет меньше порога.

------------------

Алгоритм KK:
1) Ввести и посчитать норму между узлами в графе по их расположению в графе.
2) Посчитать норму $L_2$ между узлами для итогового изображения на экране.
3) Оптимизировать их таким образом, чтобы энергия системы была минимально возможной.

Минимум есть минимум, поэтому изображение получится идентичным каждый раз, если только специально не вмешиваться в процесс и логику вывода.
